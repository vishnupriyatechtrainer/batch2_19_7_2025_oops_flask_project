```python
class Box:
    def __init__(self,length,breadth):
        self.length= length
        self.breadth=breadth
    def calculatearea(self):
        print('calculatearea',self.length+self.breadth)


box1=Box(10,30)
print(box1.length)
print(box1.breadth)
box1.calculatearea()

box2=Box(20,40)
print(box2.length)
print(box2.breadth)
box2.calculatearea()
```

```python
class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.__balance=balance # private account 
    def deposit(self,amount):
        if amount>0:
            self.balance +=amount
            print(f"deposited ${amount}.new balance is ${self.balance}")
        else:
            print('deposit amount must be greater than zero')
    def check_balance(self):
        print(f"{self.balance}")
    def get_balance(self):
        return self.__balance
    
```

```python
john_account=BankAccount("john",10000)
emma_account=BankAccount("emma",150000)


john_account.deposit(500)
john_account.check_balance()
john_account.get_balance()

emma_account.deposit(1000000)
emma_account.check_balance()
```
